```python
"""
Step a:
Author: Najnin Sultana
Assignment: #1

"""
# Step b: Create 4 variables with comments for data types
gym_member = "Alex Alliton"  # string
preferred_weight_kg = 20.5  # float
highest_reps = 25  # integer
membership_active = True  # boolean

# Step c: Create a dictionary named workout_stats to store workout statistics
# Dictionary with string keys and tuple values
workout_stats = {
    "Alex": (30, 45, 20),  # Keys: Names & values: yoga, running, weightlifting
    "Jamie": (55, 35, 35),
    "Taylor": (20, 50, 30)
}

# Step d: Calculate total workout minutes for each friend and add to dictionary
# Iterate over a copy of the dictionary items to avoid modifying it during iteration
for friend, stats in list(workout_stats.items()):  # Use list() to create a copy
    total_minutes = sum(stats)
    workout_stats[f"{friend}_Total"] = total_minutes  # Calculate total

# Step E: Create a 2-dimensional list called workout_list
# List of lists, where each inner list represents a friend's workout minutes
workout_list = [list(stats) for stats in workout_stats.values() if isinstance(stats, tuple)]

# Step F: Slice the workout_list
# Extract and print yoga and running minutes for all friends
yoga_running_minutes = [row[:2] for row in workout_list]
print("Yoga and Running Minutes for All Friends:", yoga_running_minutes)

# Extract and print weightlifting minutes for the last two friends
weightlifting_minutes = [row[2] for row in workout_list[-2:]]
print("Weightlifting Minutes for Last Two Friends:", weightlifting_minutes)

# Step G: Use an if-statement within a loop to check total workout minutes are >= 120
for friend, total in workout_stats.items():
    if isinstance(total, int) and total >= 120:
        print(f"Great job staying active, {friend.split('_')[0]}!")

# Step h: Allow user to input a friend’s name and check if it exists in the dictionary
friend_name = input("Enter a friend's name: ")
if friend_name in workout_stats:
    print(f"{friend_name}'s workout minutes: Yoga - {workout_stats[friend_name][0]}, "
          f"Running - {workout_stats[friend_name][1]}, Weightlifting - {workout_stats[friend_name][2]}")
    print(f"Total workout minutes: {workout_stats[f'{friend_name}_Total']}")
else:
    print(f"Friend {friend_name} not found in the records.")

# Step i: Print the friend with the highest and lowest total workout minutes
totals = {friend: total for friend, total in workout_stats.items() if isinstance(total, int)}
max_friend = max(totals, key=totals.get)
min_friend = min(totals, key=totals.get)
print(f"Friend with the highest total workout minutes: {max_friend} ({totals[max_friend]} minutes)")
print(f"Friend with the lowest total workout minutes: {min_friend} ({totals[min_friend]} minutes)")  
```
